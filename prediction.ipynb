```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import load_model
```

```python
word_index = imdb.get_word_index()
index_word = {v: k for k, v in word_index.items()}
```

```python
model = load_model('simplernn_imdb.h5')
model.summary()
```

```python
model.get_weights()
```

```python
def decode_review(text):
    return ' '.join([index_word.get(i - 3, '?') for i in text])
def preprocess_review(review):
    words = review.lower().split()
    encoded = [word_index.get(word, 2) + 3 for word in words]
    padded = sequence.pad_sequences([encoded], maxlen=500)
    return padded
```

```python
# Prediction function
def predict_review(review):
    processed = preprocess_review(review)
    prediction = model.predict(processed)
    sentiment = 'Positive' if prediction[0][0] >= 0.5 else 'Negative'
    return sentiment,prediction[0][0]
```

```python
example_review = "This movie was fantastic! I loved it."
predicted_sentiment, confidence = predict_review(example_review)
print(f"Review: {example_review}")
print(f"Predicted Sentiment: {predicted_sentiment}")
print(f"Confidence: {confidence}")
```
