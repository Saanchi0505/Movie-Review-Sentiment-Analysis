```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
sent = [
    "the glass of milk",
    "the glass of juice",
    "the cup of tea",
    "I am a good boy",
    "I am a good developer",
    "understand the meaning of words",
    "your videos are good"
]
```

```python
sent
```

```python
# define vocabulary size
vocab_size = 10000
```

```python
# one hot representation
one_hot_representations = [one_hot(words, vocab_size) for words in sent]
one_hot_representations
```

```python
from tensorflow.keras.layers import Embedding
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
```

```python
# padding is use to make all the sentences of same length
sent_length = 8
embedded_docs = pad_sequences(one_hot_representations,padding ='pre',maxlen=sent_length)
embedded_docs
```

```python
# feature representation using embedding
dim=10
model = Sequential()
model.add(Embedding(vocab_size,dim,input_length  = sent_length))
model.compile('adam','mse')
```

```python
model.summary()
```

```python
embedded_docs[0]
```

```python
model.predict(embedded_docs)
```
